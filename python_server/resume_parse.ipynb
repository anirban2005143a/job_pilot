```python
import pymupdf4llm
from pathlib import Path
from IPython.display import Markdown, display


# --------------------------------------------------
# Files
# --------------------------------------------------

resume_path = Path(
    r"D:\projects\job pilot\backend\uploads\resumes\6a80a9dfa56fc16bd55441ca\Anirban_Das_off_camus.pdf"
)

markdown_path = Path(
    r"D:\projects\job pilot\backend\uploads\resumes\6a80a9dfa56fc16bd55441ca\Anirban_Das_off_camus.md"
)


# --------------------------------------------------
# Validate
# --------------------------------------------------

if not resume_path.exists():
    raise FileNotFoundError(f"Resume not found: {resume_path}")


# --------------------------------------------------
# PDF → Markdown
# --------------------------------------------------

markdown = pymupdf4llm.to_markdown(
    str(resume_path)
)


# --------------------------------------------------
# Save Markdown
# --------------------------------------------------

markdown_path.write_text(
    markdown,
    encoding="utf-8"
)


# --------------------------------------------------
# Display
# --------------------------------------------------

display(Markdown(markdown))

print(f"\nMarkdown saved to:\n{markdown_path}")
```
